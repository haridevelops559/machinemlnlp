```python
import requests
from bs4 import BeautifulSoup
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Preformatted, Spacer
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.units import inch
from google.colab import files

# Wikipedia API endpoint
API_URL = "https://en.wikipedia.org/w/api.php"
headers = {"User-Agent": "CarnaticMusicBot/1.0 (https://yourdomain.com; contact@example.com)"}

# 20 raga-focused Wikipedia pages
raga_pages = [
    "Melakarta_ragas",
    "Janya_raga",
    "Raga",
    "List_of_Carnatic_ragas",
    "Arohana",
    "Svara",
    "Classification_of_ragas",
    "Thaat",
    "Gamaka",
    "Vakra_raga",
    "Audava_raga",
    "Shadava_raga",
    "Sampoorna_raga",
    "Film_songs_based_on_ragas",
    "Melakarta",
    "Raga",
    "Rasa_(aesthetics)",
    "List_of_Carnatic_ragas",
    "Alapana",
    "Tyagaraja"
]

def fetch_wikipedia_text(page_title):
    """Fetch free-form text (paragraphs) from a Wikipedia page."""
    params = {
        "action": "parse",
        "page": page_title,
        "prop": "text",
        "format": "json"
    }
    response = requests.get(API_URL, params=params, headers=headers)
    response.raise_for_status()
    html_content = response.json()["parse"]["text"]["*"]

    soup = BeautifulSoup(html_content, "html.parser")
    # Remove tables, scripts, styles, references
    for element in soup(["table", "script", "style", "sup"]):
        element.decompose()
    # Extract paragraphs
    return [p.get_text().strip() for p in soup.find_all("p") if p.get_text().strip()]

# === Create preformatted PDF ===
pdf_file = "CarnaticRaga_WikipediaPreformatted.pdf"
doc = SimpleDocTemplate(pdf_file, pagesize=letter,
                        rightMargin=inch, leftMargin=inch,
                        topMargin=inch, bottomMargin=inch)
styles = getSampleStyleSheet()
story = []

for page in raga_pages:
    # Add page title as header
    story.append(Preformatted(f"\n===== {page.replace('_', ' ')} =====\n", styles['Normal']))

    # Fetch paragraphs
    try:
        paragraphs = fetch_wikipedia_text(page)
    except Exception as e:
        print(f"Error fetching {page}: {e}")
        continue

    # Add paragraphs as preformatted text
    for para in paragraphs:
        story.append(Preformatted(para + "\n", styles['Code']))  # monospaced style
        story.append(Spacer(1, 0.1*inch))

# Build PDF
doc.build(story)
print(f"✅ Preformatted PDF saved as {pdf_file}")

# Download PDF in Colab
files.download(pdf_file)
```

```python
import os
from PyPDF2 import PdfReader

pdf_folder = "/content"  # your files folder
pdf_files = [f for f in os.listdir(pdf_folder) if f.endswith(".pdf")]

pdf_corpus = []

for pdf in pdf_files:
    reader = PdfReader(os.path.join(pdf_folder, pdf))
    for page_num, page in enumerate(reader.pages, start=1):
        text = page.extract_text()
        if text:
            pdf_corpus.append({
                "source": pdf,
                "page": page_num,
                "content": text.strip()
            })

print(f"Extracted {len(pdf_corpus)} paragraphs from {len(pdf_files)} PDFs")
```

```python


# -------------------------------
# Imports
# -------------------------------
import json
from transformers import pipeline
from sentence_transformers import SentenceTransformer, util
import torch

# -------------------------------
# Load PDF corpus JSON
# -------------------------------
with open("/content/pdf_corpus.json", "r", encoding="utf-8") as f:
    corpus_data = json.load(f)

# Adjust this based on your JSON structure
pdf_texts = corpus_data['pdf_texts']  # list of dicts with 'content' key
paragraph_texts = [p['content'] for p in pdf_texts]

print(f"Loaded {len(paragraph_texts)} paragraphs from the PDF corpus.")

# -------------------------------
# Create embeddings for semantic search
# -------------------------------
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
corpus_embeddings = embed_model.encode(paragraph_texts, convert_to_tensor=True)
print("Corpus embeddings ready.")

# -------------------------------
# Initialize Hugging Face QA pipeline
# -------------------------------
qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
print("QA pipeline ready.")

# -------------------------------
# Function: ask a question
# -------------------------------
def ask_question(question, top_k=3):
    # Embed the question
    question_embedding = embed_model.encode(question, convert_to_tensor=True)

    # Semantic search: cosine similarity
    cosine_scores = util.cos_sim(question_embedding, corpus_embeddings)[0]
    top_results = torch.topk(cosine_scores, k=top_k)

    answers = []
    for score, idx in zip(top_results.values, top_results.indices):
        context = paragraph_texts[idx]
        try:
            ans = qa_pipeline(question=question, context=context)
            answers.append({
                "answer": ans['answer'],
                "score": float(ans['score']),
                "context": context[:200]+"..."  # preview of context
            })
        except:
            continue
    return answers

# -------------------------------
# Example usage
# -------------------------------
question = "What is melakartha?"
answers = ask_question(question, top_k=2)

for i, ans in enumerate(answers):
    print(f"Answer {i+1}: {ans['answer']}")
    print(f"Score: {ans['score']:.4f}")
    print(f"Context preview: {ans['context']}\n")
```

```python
# -------------------------------
# 1️⃣ Install dependencies
# -------------------------------


# -------------------------------
# 2️⃣ Imports
# -------------------------------
import json
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer
from transformers import pipeline
from IPython.display import display
import ipywidgets as widgets

# -------------------------------
# 3️⃣ Load PDF corpus JSON
# -------------------------------
with open("/content/pdf_corpus.json", "r", encoding="utf-8") as f:
    corpus_data = json.load(f)

pdf_texts = corpus_data['pdf_texts']  # Adjust key if needed
texts = [p['content'] for p in pdf_texts]
print(f"Loaded {len(texts)} paragraphs from the PDF corpus.")

# -------------------------------
# 4️⃣ Create embeddings
# -------------------------------
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
corpus_embeddings = embed_model.encode(texts, convert_to_tensor=True)
print("Corpus embeddings ready.")

# -------------------------------
# 5️⃣ Build FAISS index
# -------------------------------
embedding_dim = corpus_embeddings.shape[1]
index = faiss.IndexFlatL2(embedding_dim)
index.add(np.array(corpus_embeddings.cpu()))
print("FAISS index built for retrieval.")

# -------------------------------
# 6️⃣ Initialize generation model (RAG-style)
# -------------------------------
generator = pipeline("text2text-generation", model="google/flan-t5-base")
print("Generation model ready.")

# -------------------------------
# 7️⃣ Define RAG query function
# -------------------------------
def rag_query(question, top_k=4, max_answer_length=300):
    # Embed the question
    q_emb = embed_model.encode(question, convert_to_tensor=True).cpu().numpy()

    # Reshape to 2D: [1, embedding_dim]
    if len(q_emb.shape) == 1:
        q_emb = q_emb.reshape(1, -1)

    # Retrieve top-k relevant paragraphs
    D, I = index.search(q_emb, top_k)
    retrieved_texts = [texts[i] for i in I[0]]

    # Combine retrieved paragraphs as context
    context = " ".join(retrieved_texts)

    # Prepare input for the generator
    input_text = f"Context: {context}\nQuestion: {question}\nAnswer:"

    # Generate answer
    output = generator(input_text, max_length=max_answer_length, do_sample=False)
    return output[0]['generated_text']

# -------------------------------
# 8️⃣ Interactive Widgets
# -------------------------------
question_box = widgets.Text(
    description='Question:',
    placeholder='Type your question here',
    layout=widgets.Layout(width='80%')
)

output_box = widgets.Output()

def on_submit(change):
    with output_box:
        output_box.clear_output()
        question = change['new']
        if question.strip():
            answer = rag_query(question, top_k=3)
            print("Question:", question)
            print("Answer:", answer)

question_box.observe(on_submit, names='value')

display(question_box, output_box)
```

```python
import os
import zipfile
import pandas as pd

# 1. Extract all zip files into their own folders
zip_files = [f for f in os.listdir() if f.endswith(".zip")]
for z in zip_files:
    folder_name = z.replace(".zip", "")
    if not os.path.exists(folder_name):
        os.makedirs(folder_name, exist_ok=True)
        with zipfile.ZipFile(z, 'r') as zip_ref:
            zip_ref.extractall(folder_name)
        print(f"✅ Extracted {z} into {folder_name}/")

# 2. Collect all CSVs (both loose + inside extracted folders)
csv_files = []
for root, _, files in os.walk("."):  # search recursively
    for f in files:
        if f.endswith(".csv"):
            csv_files.append(os.path.join(root, f))

print(f"\n📂 Found {len(csv_files)} CSV files in total")

# 3. Convert rows to dictionaries (with source info)
csv_corpus = []
for csv in csv_files:
    try:
        df = pd.read_csv(csv)
        for _, row in df.iterrows():
            row_dict = row.to_dict()
            row_dict["source"] = os.path.basename(csv)  # keep track of file name
            csv_corpus.append(row_dict)
    except Exception as e:
        print(f"⚠️ Skipped {csv} due to error: {e}")

print(f"\n✅ Extracted {len(csv_corpus)} rows from {len(csv_files)} CSVs")

# Optional: Save full corpus as JSON for NLP usage
import json
with open("carnatic_csv_corpus.json", "w") as f:
    json.dump(csv_corpus, f, indent=2)

print("📄 Saved merged corpus as carnatic_csv_corpus.json")
```

```python
import json
import re

# -------------------------
# Load your unified corpus
# -------------------------
with open("carnatic_master_corpus.json", "r") as f:
    master_corpus = json.load(f)

# -------------------------
# Function to normalize text
# -------------------------
def normalize_text(text):
    if not text:
        return ""
    text = text.lower()  # lowercase
    text = re.sub(r'[^a-z0-9\s]', '', text)  # remove special characters
    text = re.sub(r'\s+', ' ', text).strip()  # collapse multiple spaces
    return text

# -------------------------
# Normalize PDF texts
# -------------------------
for entry in master_corpus.get("pdf_texts", []):
    entry["content"] = normalize_text(entry.get("content", ""))

# -------------------------
# Normalize CSV rows
# -------------------------
for row in master_corpus.get("csv_rows", []):
    for key, value in row.items():
        if isinstance(value, str):
            row[key] = normalize_text(value)

# -------------------------
# Save normalized corpus
# -------------------------
with open("carnatic_master_corpus_normalized.json", "w") as f:
    json.dump(master_corpus, f, indent=2)

print("✅ Text normalization complete and saved as carnatic_master_corpus_normalized.json")
```

```python
import spacy
import json

nlp = spacy.load("en_core_web_sm")  # load small English model

# Load normalized corpus
with open("carnatic_master_corpus_normalized.json", "r") as f:
    master_corpus = json.load(f)

# Function for stopword removal + lemmatization
def preprocess_spacy(text):
    doc = nlp(text)
    tokens = [token.lemma_ for token in doc if not token.is_stop and not token.is_punct]
    return " ".join(tokens)

# Apply only to PDF texts
for entry in master_corpus.get("pdf_texts", []):
    entry["content_processed"] = preprocess_spacy(entry.get("content", ""))

# Save processed corpus
with open("carnatic_master_corpus_pdf_processed.json", "w") as f:
    json.dump(master_corpus, f, indent=2)

print("✅ Stopword removal & lemmatization applied to PDF corpus using spaCy")
```

```python
import json

# -----------------------------
# Load merged CSV corpus
# -----------------------------
with open("carnatic_csv_corpus.json", "r") as f:
    csv_corpus = json.load(f)

# -----------------------------
# Define canonical schema mapping
# -----------------------------
schema_mapping = {
    "raga_name": ["Raga Name", "Ragam", "Raga", "raga_name"],
    "arohana": ["Ārōhaṇa", "Arohana", "arohana"],
    "avarohana": ["Avarōhaṇa", "Avarohana", "avarohana"],
    "composer": ["Composer", "Sangeeta Karta", "composer"],
    "tala": ["Tala", "Talam", "tala"]
}

# -----------------------------
# Function to align row keys to canonical schema
# -----------------------------
def align_row_schema(row, schema_mapping):
    aligned_row = {}
    for canonical_key, variants in schema_mapping.items():
        for variant in variants:
            if variant in row:
                aligned_row[canonical_key] = row[variant]
                break
        # If no variant found in this row, set value as None
        if canonical_key not in aligned_row:
            aligned_row[canonical_key] = None
    # Keep other columns not in canonical schema as-is (optional)
    for col, val in row.items():
        if col not in sum(schema_mapping.values(), []):
            aligned_row[col] = val
    return aligned_row

# -----------------------------
# Apply alignment to all rows
# -----------------------------
aligned_csv_rows = [align_row_schema(row, schema_mapping) for row in csv_corpus]

# -----------------------------
# Save aligned CSV corpus
# -----------------------------
with open("carnatic_csv_corpus_aligned.json", "w") as f:
    json.dump(aligned_csv_rows, f, indent=2)

print(f"✅ Aligned CSV corpus saved with {len(aligned_csv_rows)} rows as carnatic_csv_corpus_aligned.json")
```

```python
import requests
from bs4 import BeautifulSoup
import json

def fetch_wiki_glossary(page_title="Glossary_of_Carnatic_music"):
    # Wikipedia API endpoint
    API_URL = "https://en.wikipedia.org/w/api.php"
    params = {
        "action": "parse",
        "page": page_title,
        "prop": "text",
        "format": "json"
    }
    headers = {
        "User-Agent": "CarnaticCorpusBot/1.0 (https://yourdomain.com; contact@example.com)"
    }

    # Fetch page content
    resp = requests.get(API_URL, params=params, headers=headers)
    resp.raise_for_status()

    data = resp.json()
    html_content = data["parse"]["text"]["*"]

    # Parse HTML
    soup = BeautifulSoup(html_content, "html.parser")

    # Remove unwanted tags
    for element in soup(["table", "script", "style", "sup", "img"]):
        element.decompose()

    # Extract terms and definitions
    glossary = {}
    # Many Wikipedia glossary pages use bold term names at start of paragraphs or list items
    # So try to parse <p> tags
    for p in soup.find_all("p"):
        text = p.get_text().strip()
        if not text:
            continue
        # If the paragraph starts with a bold term
        bold = p.find("b")
        if bold:
            term = bold.get_text().strip()
            # The rest of the paragraph as definition (minus the term)
            def_text = text[len(term):].strip(" –—:;,.")
            glossary[term] = def_text
        else:
            # Optional: fallback, parse if colon appears
            if ":" in text:
                parts = text.split(":", 1)
                term = parts[0].strip()
                def_text = parts[1].strip()
                # Only accept if term is fairly short
                if len(term.split()) <= 3:
                    glossary[term] = def_text

    return glossary

def merge_into_unified(unified_filepath="carnatic_unified_corpusfinal.json",
                        output_filepath="carnatic_corpus_best.json"):
    # 1. Load existing unified corpus
    with open(unified_filepath, "r", encoding="utf-8") as f:
        unified = json.load(f)

    # 2. Fetch glossary
    glossary = fetch_wiki_glossary()

    # 3. Add the glossary as a new section
    unified["wiki_glossary"] = glossary

    # 4. Save merged corpus
    with open(output_filepath, "w", encoding="utf-8") as f:
        json.dump(unified, f, ensure_ascii=False, indent=2)

    print(f"✅ Merged glossary added. Saved to {output_filepath}")
    print(f"Number of glossary terms added: {len(glossary)}")

# —— Run the merge
merge_into_unified("carnatic_unified_corpusfinal.json", "carnatic_corpus_best.json")
```

```python
def lookup_raga_info(query: str):
    """
    Simple lookup for arohana/avarohana, tala, composer.
    """
    q = query.lower()
    for raga_name, info in raga_dict.items():
        if raga_name.lower() in q:
            return {
                "raga": raga_name,
                "arohana": info.get("arohana"),
                "avarohana": info.get("avarohana"),
                "tala": info.get("tala"),
                "composer": info.get("composer")
            }
    return None
```

```python
from transformers import pipeline

# Extractive QA
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

# Summarisation
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
```

```python
def hybrid_query(user_query: str, top_k=3):
    # 1. Try structured lookup (direct JSON/dictionary)
    structured = lookup_raga_info(user_query)
    if structured:
        return {"type": "structured", "answer": structured}

    # 2. Else → freeform retrieval + QA/summarisation
    retrieved = retrieve_pdf_chunks(user_query, top_k=top_k)
    context = " ".join([r["sentence"] for r in retrieved])

    # Try QA
    try:
        qa_result = qa_pipeline(question=user_query, context=context)
        answer_text = qa_result.get("answer", "")
    except:
        answer_text = ""

    # If QA fails, fallback to summarisation
    if not answer_text or answer_text.strip() == "":
        summary = summarizer(context, max_length=80, min_length=30, do_sample=False)
        answer_text = summary[0]['summary_text']

    return {
        "type": "freeform",
        "answer": answer_text,
        "retrieved_sentences": retrieved
    }
```

```python
from sentence_transformers import SentenceTransformer, util

# 1️⃣ Prepare embeddings once
model = SentenceTransformer("all-MiniLM-L6-v2")  # Fast & effective

# Flatten PDF corpus sentences
sentence_store = []
for entry in pdf_corpus.get("pdf_texts", []):
    for sent_info in entry.get("enriched_sentences", []):
        sentence_store.append(sent_info)

# Compute embeddings for all sentences
corpus_sentences = [s["sentence"] for s in sentence_store]
corpus_embeddings = model.encode(corpus_sentences, convert_to_tensor=True)

# 2️⃣ Smarter retrieval function
def retrieve_paragraph(user_query, top_k=10, merge_window=2):
    """
    Retrieves top_k sentences semantically similar to the query,
    then merges neighboring sentences into paragraph-style output.
    """
    query_emb = model.encode(user_query, convert_to_tensor=True)
    scores = util.cos_sim(query_emb, corpus_embeddings)[0]

    # Get top_k sentence indices
    top_results = scores.topk(k=top_k)
    indices = top_results[1].cpu().numpy()

    # Merge neighboring sentences to form paragraphs
    merged_paragraphs = []
    last_idx = -merge_window - 1
    current_para = []
    for idx in sorted(indices):
        if idx - last_idx <= merge_window:
            current_para.append(corpus_sentences[idx])
        else:
            if current_para:
                merged_paragraphs.append(" ".join(current_para))
            current_para = [corpus_sentences[idx]]
        last_idx = idx
    if current_para:
        merged_paragraphs.append(" ".join(current_para))

    return merged_paragraphs
```

```python
def hybrid_query_paragraph(user_query: str, top_k=10):
    """
    Hybrid query returning structured info + paragraph-style freeform explanation.
    """
    user_query_lower = user_query.lower()
    structured_results = []

    # 1️⃣ Structured dictionary lookup
    for raga_name, info in raga_dict.items():
        if raga_name.lower() in user_query_lower:
            structured_results.append({
                "raga": raga_name,
                "arohana": info.get("arohana"),
                "avarohana": info.get("avarohana"),
                "tala": info.get("tala"),
                "composer": info.get("composer")
            })

    # 2️⃣ Paragraph-style PDF retrieval
    paragraphs = retrieve_paragraph(user_query, top_k=top_k)

    # 3️⃣ Optional QA or summarization (if you want to condense)
    full_context = " ".join(paragraphs)
    answer_text = ""
    if full_context.strip():
        try:
            qa_result = qa_pipeline(question=user_query, context=full_context)
            answer_text = qa_result.get("answer", "")
        except:
            answer_text = ""

    # 4️⃣ Fallback summarization to make coherent paragraph
    if not answer_text.strip() and full_context.strip():
        summary = summarizer(full_context, max_length=200, min_length=80, do_sample=False)
        answer_text = summary[0]['summary_text']

    return {
        "structured": structured_results if structured_results else None,
        "paragraphs": paragraphs,
        "answer_text": answer_text
    }
```

```python
import json
import torch
from sentence_transformers import SentenceTransformer, util
from transformers import pipeline

# -----------------------------
# Step 1: Load corpora
# -----------------------------
with open("carnatic_master_corpus_pdf_sentences.json", "r") as f:
    pdf_corpus = json.load(f)

with open("carnatic_csv_corpus_aligned.json", "r") as f:
    csv_corpus = json.load(f)

with open("raga_dict_cleaned.json", "r") as f:
    raga_dict = json.load(f)

# -----------------------------
# Step 2: Initialize models
# -----------------------------
embedder = SentenceTransformer('all-MiniLM-L6-v2')  # 384-d embeddings
qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# -----------------------------
# Step 3: Prepare PDF sentences
# -----------------------------
pdf_sentences = []
pdf_sentence_to_entry = []

for entry in pdf_corpus.get("pdf_texts", []):
    entry_sentences = entry.get("sentences", [])
    for s in entry_sentences:
        if isinstance(s, dict) and "sentence" in s:
            pdf_sentences.append(s["sentence"])
            pdf_sentence_to_entry.append(entry)
        elif isinstance(s, str):
            pdf_sentences.append(s)
            pdf_sentence_to_entry.append(entry)

pdf_embeddings = embedder.encode(pdf_sentences, convert_to_tensor=True) if pdf_sentences else None

# -----------------------------
# Step 4: Build CSV reverse lookup
# -----------------------------
# Map values to rows for structured retrieval
value_to_rows = {}
for row in csv_corpus:
    for col, val in row.items():
        if val and isinstance(val, str):
            key = val.lower().strip()
            if key not in value_to_rows:
                value_to_rows[key] = []
            value_to_rows[key].append(row)

# -----------------------------
# Step 5: Helper functions
# -----------------------------
def retrieve_pdf_chunks(query, top_k=5):
    if pdf_embeddings is None or len(pdf_sentences) == 0:
        return []

    query_emb = embedder.encode(query, convert_to_tensor=True)
    cos_scores = util.cos_sim(query_emb, pdf_embeddings)[0]

    # Handle the case when pdf_embeddings is empty
    if cos_scores.numel() == 0:
        return []

    top_results = torch.topk(cos_scores, k=min(top_k, len(pdf_sentences)))

    retrieved = []
    for score, idx in zip(top_results.values, top_results.indices):
        idx_int = idx.item()
        if idx_int < len(pdf_sentences):
            retrieved.append({
                "sentence": pdf_sentences[idx_int],
                "score": score.item()
            })
    return retrieved


def lookup_raga_info(user_query):
    results = []
    user_query_lower = user_query.lower()
    for raga_name, info in raga_dict.items():
        if raga_name.lower() in user_query_lower:
            results.append({
                "raga": raga_name,
                "arohana": info.get("arohana"),
                "avarohana": info.get("avarohana"),
                "tala": info.get("tala"),
                "composer": info.get("composer")
            })
    return results if results else None

def lookup_csv_info(user_query):
    """
    Return structured rows from CSV if any cell matches a keyword in query.
    """
    results = []
    query_words = [w.lower() for w in user_query.split()]

    for word in query_words:
        if word in value_to_rows:
            for row in value_to_rows[word]:
                if row not in results:
                    results.append(row)
    return results if results else None

# -----------------------------
# Step 6: Full hybrid pipeline
# -----------------------------
def hybrid_carnatic_query(user_query, top_k=5):
    """
    Returns combined:
      - Raga dictionary info
      - CSV structured info (composer, tala, arohana, avarohana, song)
      - PDF freeform explanations
      - QA + summarization fallback
    """
    result = {
        "raga_info": None,
        "csv_info": None,
        "freeform": None,
        "retrieved_sentences": []
    }

    # 1️⃣ Structured lookup
    raga_info = lookup_raga_info(user_query)
    csv_info = lookup_csv_info(user_query)
    if raga_info:
        result["raga_info"] = raga_info
    if csv_info:
        result["csv_info"] = csv_info

    # 2️⃣ PDF retrieval
    retrieved = retrieve_pdf_chunks(user_query, top_k=top_k)
    result["retrieved_sentences"] = retrieved
    context_text = " ".join([r["sentence"] for r in retrieved])

    # 3️⃣ QA over PDF context
    answer_text = ""
    if context_text.strip():
        try:
            qa_result = qa_pipeline(question=user_query, context=context_text)
            answer_text = qa_result.get("answer", "")
        except:
            answer_text = ""

    # 4️⃣ Fallback summarization
    if not answer_text.strip() and context_text.strip():
        try:
            summary = summarizer(context_text, max_length=150, min_length=60, do_sample=False)
            answer_text = summary[0]['summary_text']
        except:
            answer_text = ""

    # 5️⃣ If QA/summarization fails, join retrieved sentences
    if not answer_text.strip() and retrieved:
        answer_text = " ".join([r["sentence"] for r in retrieved])

    result["freeform"] = answer_text if answer_text else None
    return result

# -----------------------------
# Step 7: Example usage
# -----------------------------
query = "Explain kritis composed by Thyagaraja and their ragas"
res = hybrid_carnatic_query(query, top_k=5)

print("\n--- Raga Dictionary Info ---")
if res["raga_info"]:
    for r in res["raga_info"]:
        print(r)

print("\n--- CSV Structured Info ---")
if res["csv_info"]:
    for r in res["csv_info"]:
        print(r)

print("\n--- Freeform Explanation / Paragraph ---")
if res["freeform"]:
    print(res["freeform"])

print("\n--- Retrieved Sentences ---")
for r in res["retrieved_sentences"]:
    print("-", r["sentence"])
```

```python
import json
from sentence_transformers import SentenceTransformer
from langchain.vectorstores import FAISS
from langchain.embeddings import HuggingFaceEmbeddings
from transformers import pipeline

# -----------------------------
# Step 1: Load corpora
# -----------------------------
with open("raga_dict_cleaned.json", "r") as f:
    raga_dict = json.load(f)

with open("carnatic_csv_corpus_aligned.json", "r") as f:
    csv_corpus = json.load(f)

with open("carnatic_master_corpus_pdf_sentences.json", "r") as f:
    pdf_corpus = json.load(f)

# -----------------------------
# Step 2: Embedding & FAISS
# -----------------------------
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Flatten PDF sentences
pdf_sentences = []
for entry in pdf_corpus.get("pdf_texts", []):
    for s in entry.get("sentences", []):
        if isinstance(s, dict):
            pdf_sentences.append(s.get("sentence", ""))
        else:
            pdf_sentences.append(s)

# Build FAISS index
db = FAISS.from_texts(pdf_sentences, embeddings)

# -----------------------------
# Step 3: NLP models
# -----------------------------
qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# -----------------------------
# Step 4: Lookup functions
# -----------------------------
def lookup_raga_info(query):
    q = query.lower()
    matches = []
    for raga, info in raga_dict.items():
        if raga.lower() in q:
            matches.append({
                "raga": raga,
                "arohana": info.get("arohana"),
                "avarohana": info.get("avarohana"),
                "tala": info.get("tala"),
                "composer": info.get("composer")
            })
    return matches if matches else None

def lookup_csv_info(query):
    q = query.lower()
    matches = []
    for row in csv_corpus:
        if any(val and isinstance(val, str) and val.lower() in q for val in row.values()):
            matches.append(row)
    return matches if matches else None

# -----------------------------
# Step 5: Hybrid Query
# -----------------------------
def hybrid_query(query, top_k=5):
    result = {
        "structured_raga": lookup_raga_info(query),
        "structured_csv": lookup_csv_info(query),
        "freeform_answer": None,
        "retrieved_sentences": []
    }

    # Retrieve context from FAISS
    docs = db.similarity_search(query, k=top_k)
    context = " ".join([d.page_content for d in docs])
    result["retrieved_sentences"] = [d.page_content for d in docs]

    # Run QA
    answer = ""
    if context.strip():
        try:
            qa_res = qa_pipeline(question=query, context=context)
            answer = qa_res.get("answer", "")
        except:
            answer = ""

    # Fallback to summarization (multi-sentence answer)
    if not answer or len(answer.split()) < 5:
        try:
            summary = summarizer(context, max_length=150, min_length=60, do_sample=False)
            answer = summary[0]["summary_text"]
        except:
            answer = context

    result["freeform_answer"] = answer
    return result

# -----------------------------
# Step 6: Example Queries
# -----------------------------
queries = [
    "What is the arohana of Mohanam?",
    "Explain about kritis in Carnatic music",
    "List composers of popular ragas"
]

for q in queries:
    print(f"\n🔍 Query: {q}")
    res = hybrid_query(q, top_k=5)

    if res["structured_raga"]:
        print("🎼 Raga Info:", res["structured_raga"])
    if res["structured_csv"]:
        print("📊 CSV Info:", res["structured_csv"][:2])  # preview first 2 rows
    print("📖 Freeform Answer:", res["freeform_answer"])
```

```python
import json
from difflib import SequenceMatcher

# Load raga_dict
with open("raga_dict_cleaned.json", "r") as f:
    raga_dict = json.load(f)

# -------------------------------
# Function to compute similarity between two scales
# -------------------------------
def scale_similarity(scale1, scale2):
    """
    Compare two swara sequences and return similarity ratio.
    Uses difflib.SequenceMatcher to allow minor differences.
    """
    if not scale1 or not scale2:
        return 0.0
    # Normalize: remove spaces
    s1 = scale1.replace(" ", "")
    s2 = scale2.replace(" ", "")
    return SequenceMatcher(None, s1, s2).ratio()

# -------------------------------
# Input sample raga
# -------------------------------
sample_raga_name = "Mohanam"  # replace with any raga name
sample_raga = raga_dict.get(sample_raga_name)
if not sample_raga:
    raise ValueError(f"Raga {sample_raga_name} not found in dictionary.")

sample_arohana = sample_raga.get("arohana")
sample_avarohana = sample_raga.get("avarohana")

# -------------------------------
# Threshold for similarity (0.8 = ~80% similar)
# -------------------------------
threshold = 0.9

# -------------------------------
# Find similar ragas
# -------------------------------
similar_ragas = []

for raga_name, info in raga_dict.items():
    if raga_name == sample_raga_name:
        continue
    aro_sim = scale_similarity(sample_arohana, info.get("arohana"))
    ava_sim = scale_similarity(sample_avarohana, info.get("avarohana"))

    # If both arohana and avarohana are similar enough
    if aro_sim >= threshold and ava_sim >= threshold:
        similar_ragas.append({
            "raga": raga_name,
            "arohana": info.get("arohana"),
            "avarohana": info.get("avarohana"),
            "similarity_arohana": round(aro_sim, 2),
            "similarity_avarohana": round(ava_sim, 2)
        })

# -------------------------------
# Print results
# -------------------------------
print(f"Ragas similar to {sample_raga_name}: {len(similar_ragas)} found\n")
for r in similar_ragas:
    print(f"{r['raga']}")
    print(f"  Arohana: {r['arohana']} (sim={r['similarity_arohana']})")
    print(f"  Avarohana: {r['avarohana']} (sim={r['similarity_avarohana']})\n")
```

```python
import json
from sentence_transformers import SentenceTransformer, util
from transformers import pipeline

# -------------------------------
# 1️⃣ Load raga_dict (CSV-based info)
# -------------------------------
with open("raga_dict_with_melakarta.json", "r") as f:
    raga_dict = json.load(f)

# -------------------------------
# 2️⃣ Load PDF sentences with NER
# -------------------------------
with open("carnatic_master_corpus_pdf_ner.json", "r") as f:
    pdf_corpus = json.load(f)

# -------------------------------
# 3️⃣ Build a list of sentences with Raga references
# -------------------------------
pdf_sentences = []
pdf_sentence_raga_map = []
for entry in pdf_corpus.get("pdf_texts", []):
    for sent_data in entry.get("entities", []):
        sentence = sent_data["sentence"]
        entities = sent_data["entities"]
        if "RAGA" in entities:
            pdf_sentences.append(sentence)
            pdf_sentence_raga_map.append({
                "sentence": sentence,
                "raga_entities": entities.get("RAGA", []),
                "tala_entities": entities.get("TALAM", []),
                "composition_entities": entities.get("COMPOSITION", [])
            })

print(f"Total PDF sentences mentioning RAGAs: {len(pdf_sentences)}")

# -------------------------------
# 4️⃣ Encode sentences using SentenceTransformer
# -------------------------------
model = SentenceTransformer('all-MiniLM-L6-v2')
sentence_embeddings = model.encode(pdf_sentences, convert_to_tensor=True)

# -------------------------------
# 5️⃣ Load summarization pipeline
# -------------------------------
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# -------------------------------
# 6️⃣ Query function with summarization
# -------------------------------
def query_raga_summary(user_query, top_k=10, max_paragraph_chars=2000):
    results = {}
    raga_name = user_query.strip()

    # ---------------- Exact match from raga_dict ----------------
    if raga_name in raga_dict:
        info = raga_dict[raga_name]
        results["raga_info"] = {
            "arohana": info.get("arohana"),
            "avarohana": info.get("avarohana"),
            "melakarta_number": info.get("melakarta_number"),
            "songs": info.get("songs", [])
        }

    # ---------------- Semantic search in PDF sentences ----------------
    query_embedding = model.encode(user_query, convert_to_tensor=True)
    hits = util.semantic_search(query_embedding, sentence_embeddings, top_k=top_k)[0]

    # Collect relevant sentences
    relevant_sentences = [pdf_sentence_raga_map[hit["corpus_id"]]["sentence"] for hit in hits]

    # Concatenate sentences to form a paragraph for summarization
    paragraph_text = " ".join(relevant_sentences)
    if len(paragraph_text) > max_paragraph_chars:
        paragraph_text = paragraph_text[:max_paragraph_chars]  # truncate if too long

    # ---------------- Summarize ----------------
    if paragraph_text:
        summary = summarizer(paragraph_text, max_length=200, min_length=50, do_sample=False)[0]["summary_text"]
    else:
        summary = "No relevant paragraphs found for summarization."

    results["summary"] = summary
    results["pdf_sentences"] = relevant_sentences
    return results

# -------------------------------
# 7️⃣ Example usage
# -------------------------------
user_query = "Mohanam"
results = query_raga_summary(user_query, top_k=15)

# Display exact CSV info
if "raga_info" in results:
    info = results["raga_info"]
    print(f"\n=== Raga: {user_query} ===")
    print("Arohana:", info["arohana"])
    print("Avarohana:", info["avarohana"])
    print("Melakarta Number:", info["melakarta_number"])
    print("Songs:")
    for song in info["songs"]:
        print(f"  - {song['title']} by {song['composer']} (Movie: {song['movie']}, Singers: {', '.join(song['singers'])})")

# Display summarization
print("\n=== Summary of PDF Theory ===")
print(results["summary"])
```

```python
import requests
from bs4 import BeautifulSoup

def fetch_compositions(raga_name):
    # Construct the URL for the raga
    url = f"https://www.karnatik.com/ragasa.shtml#{raga_name}"

    # Send a GET request to fetch the page content
    response = requests.get(url)
    if response.status_code != 200:
        print(f"Failed to retrieve {raga_name} compositions.")
        return []

    # Parse the page content
    soup = BeautifulSoup(response.text, 'html.parser')

    # Find the section corresponding to the raga
    raga_section = soup.find(id=raga_name)
    if not raga_section:
        print(f"No compositions found for {raga_name}.")
        return []

    # Extract compositions listed under the raga
    compositions = []
    for item in raga_section.find_all_next('li'):
        composition = item.get_text(strip=True)
        if composition:
            compositions.append(composition)

    return compositions

# Example usage
raga_name = "AbhOgi"
compositions = fetch_compositions(raga_name)
if compositions:
    print(f"Top compositions for {raga_name}:")
    for comp in compositions:
        print(f"- {comp}")
else:
    print(f"No compositions found for {raga_name}.")
```

```python
import requests
from bs4 import BeautifulSoup

BASE_URL = "https://www.karnatik.com/ragas.shtml"

# -------------------------------
# 1️⃣ Build raga -> anchor mapping
# -------------------------------
def get_raga_mapping():
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(BASE_URL, headers=headers)
    response.raise_for_status()
    soup = BeautifulSoup(response.text, "html.parser")

    raga_mapping = {}
    for link in soup.find_all("a", href=True):
        href = link['href']
        name = link.get_text(strip=True)
        if href.startswith("#") and name:
            raga_mapping[name.lower()] = href  # lowercase for easy lookup
    return raga_mapping, soup

# -------------------------------
# 2️⃣ Fetch compositions for a raga
# -------------------------------
def fetch_top_compositions(raga_name, max_results=10):
    raga_name_lower = raga_name.lower()
    raga_mapping, soup = get_raga_mapping()

    if raga_name_lower not in raga_mapping:
        print(f"Raga '{raga_name}' not found in master list.")
        return []

    anchor = raga_mapping[raga_name_lower]
    # Find the section with compositions
    raga_section = soup.find("a", {"name": anchor[1:]})  # remove #
    if not raga_section:
        print(f"Section for raga '{raga_name}' not found.")
        return []

    # Compositions are usually in the following <ul> or <table>
    compositions = []
    for sibling in raga_section.find_next_siblings():
        if sibling.name == "ul":  # compositions are often <li> in <ul>
            for li in sibling.find_all("li"):
                title = li.get_text(strip=True)
                compositions.append(title)
                if len(compositions) >= max_results:
                    return compositions
            break  # stop after first <ul>
        elif sibling.name == "h2":  # next raga section
            break

    return compositions

# -------------------------------
# Example usage
# -------------------------------
raga_name = "AbhOgi"
top_songs = fetch_top_compositions(raga_name, max_results=10)
print(f"Top compositions for {raga_name}:")
for i, song in enumerate(top_songs, 1):
    print(f"{i}. {song}")
```
